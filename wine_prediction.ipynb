```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
```

```python
df = pd.read_csv(r"C:\Users\Hp\my_projects\WineQT.csv")
df
```

```python
df.isnull()
```

```python
def classify_quality(score):
    if score<=5:
        return 'Bad'
    elif score==6:
        return 'Medium'
    else:
        return 'Good'
df['quality_class'] = df['quality'].apply(classify_quality)
```

```python
X = df.drop(columns = ['quality','Id','quality_class'],axis = 1)
y = df['quality']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
```

```python
model = RandomForestClassifier()
model.fit(X_train,y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score,classification_report
```

```python
accuracy_score(y_test,y_pred)
```

```python
print(classification_report(y_test,y_pred))
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

# 1. Assuming X and y are already defined from your previous steps
# X = df.drop(['quality', 'quality_class'], axis=1)
# y = df['quality_class']

# 2. Split data before scaling
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

# 3. Create and fit the scaler on the training data only
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)

# 4. Apply the *fitted* scaler to the test data
X_test_scaled = scaler.transform(X_test)

# 5. Re-train and re-evaluate the model
model = RandomForestClassifier(random_state=42)
model.fit(X_train_scaled, y_train)

y_pred = model.predict(X_test_scaled)
print("New Accuracy:", accuracy_score(y_test, y_pred))
```
